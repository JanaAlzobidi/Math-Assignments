```python
# Import the NumPy library to work with arrays and matrices.
import numpy as np
```

```python
# Created the dataset containing information about 4 customers.
# Each row represents one customer.
# The three columns represent: Age - Number of Purchases - Total Spending in (SAR).

customer_data = np.array([
    [25, 4, 1200], # Customer 1: Age 25, 4 purchases, spent 1200 SAR.
    [32, 7, 2500],
    [28, 3, 800],
    [40, 10, 4000]
])
```

Part 1: NumPy & Math Foundations

```python
# Define examples of: Scalar, Vector, Matrix, Tensor.
# Show the shape and dimensions.

# 1. Scalar: A single numerical value.
scalar = np.array(30)
print("Scalar:", scalar)
print("Shape:", scalar.shape)
print("Dimensions:", scalar.ndim)
print("----------------------------")

# 2. Vector: A one-dimensional array containing multiple values.
vector = np.array([25, 32, 28, 40])
print("Vector:", vector)
print("Shape:", vector.shape)
print("Dimensions:", vector.ndim)
print("----------------------------")

# 3. Matrix: A two-dimensional array organized into rows and columns.
print("Matrix:\n", customer_data)
print("Shape:", customer_data.shape)
print("Dimensions:", customer_data.ndim)
print("----------------------------")

# 4. Tensor: A multi-dimensional array with three or more dimensions.
customer_tensor = np.array([
    [   # January:
        [25, 4, 1200],
        [32, 7, 2500]
    ],
    [   # February:
        [28, 3, 800],
        [40, 10, 4000]
    ]
])

print("Tensor:\n", customer_tensor)
print("Shape:", customer_tensor.shape)
print("Dimensions:", customer_tensor.ndim)
```

**Explain the difference between: (n,) and (n,1):**

(n,): A one-dimensional array containing n elements.

(n,1): A two-dimensional array containing n rows and one column.

Both contain the same number of elements, but they have different shapes and dimensions.

```python
# Apply the element-wise operations.

# Two arrays for customer spending across two months.
january_spending = np.array([1200, 2500, 800, 4000])
february_spending = np.array([1400, 2200, 1000, 4500])

print("Addition:", january_spending + february_spending)

print("Subtraction:", january_spending - february_spending)

print("Multiplication:", january_spending * february_spending)

print("Division:", january_spending / february_spending)
```

```python
# Apply aggregations.

# Extract the total spending column from the customer data matrix.
spending = customer_data[:, 2]
print(spending)

print("Sum:", np.sum(spending))

print("Mean:", np.mean(spending))

print("Min:", np.min(spending))

print("Max:", np.max(spending))

print("Standard Deviation:", np.std(spending))
```

Part 2: Vectors & Linear Algebra

```python
# Create two vectors representing customer spending in SAR.

v = np.array([1200, 2500, 800, 4000])
w = np.array([1400, 2200, 1000, 4500])

# Compute:
print("Vector Addition:", v + w)

print("Scalar multiplication:", 2 * v)

print("Dot product:", np.dot(v,w))

print("Norm:", np.linalg.norm(v))

cos_sim = np.dot(v,w) / (np.linalg.norm(v) * np.linalg.norm(w))
print("Cosine Similarity:", cos_sim)
```

```python
# Create row vector and column vector.

row_vector = np.array([[1200, 2500, 800, 4000]])
print("Row Vector:\n", row_vector)
print("Shape:", row_vector.shape)
print("----------------------------")

column_vector = np.array([[1200], [2500], [800], [4000]])
print("Column Vector:\n", column_vector)
print("Shape:", column_vector.shape)
```

**Explain the meaning of dot product: positive - zero - negative.**

Positive: The angle between the vectors is less than 90 degrees. For example, in our case, the dot product is 25,980,000.

Zero: The vectors are perpendicular to each other, forming a 90-degree angle.

Negative: The angle between the vectors is greater than 90 degrees.

```python
# Create two vectors and test:
# o Are they linearly independent?
# o Do they form a basis for R²?

v = np.array([2, 1])
w = np.array([1, 3])

A = np.column_stack([v, w])

print("Rank =", np.linalg.matrix_rank(A))
```

**Linearly independent?** Yes, because the vectors have different directions and rank = 2.

**Do they form a basis for R²?** Yes, because the vectors are linearly independent and span R².

```python
# Create a linear combination.

a = 2
b = 3

result = a * v + b * w

print("Linear Combination:", result)
```

```python
# Apply a linear transformation using a matrix.

# Scaling:
v = np.array([2, 1])

A = np.array([[2, 0],
              [0, 2]])

result = A @ v

print("Scaling:", result)

# Rotation or shear
R = np.array([[0, -1],
              [1, 0]])

result = R @ v

print("Rotation:", result)
```

**Why are vectors, matrices, and transformations important in Machine Learning?**

Vectors represent data, matrices organize data, and transformations change data to help Machine Learning models learn patterns, make predictions, and solve problems.
