### **Task 2: Multiply the Matrix by v2**

```python
import numpy as np

A = np.array([[2, 1],
              [1, 2]])

v1 = np.array([1, 1])
v2 = np.array([1, -1])

print("A @ v1 =", A @ v1)
print("A @ v2 =", A @ v2)
```

### **Task 3: Compare the Direction Before and After Transformation**
### **Task 4: Determine Which Vectors Remain on the Same Direction**

- The vector \(v_1\) keeps the **same direction** after transformation, but its **magnitude increases** by a factor of 3.

- The vector \(v_2\) remains unchanged because it is multiplied by 1, so **both its direction and magnitude stay the same**.

Both \(v_1\) and \(v_2\) remain on the same line after the transformation. Therefore, **both vectors preserve their direction**.

### **Task 5: Explain Why These Vectors Are Called Eigenvectors**

These vectors are called eigenvectors because multiplying the matrix by each vector changes only its magnitude, without changing its direction.

Mathematically,

$$
Av = \lambda v
$$

where $\lambda$ is the eigenvalue.

### **AI Insight: Why Do Most Vectors Change Direction?**

Most vectors change direction because a matrix transformation can affect their components differently.

### **Task 5: Verify the Eigenvalues Using NumPy**

```python
import numpy as np

A = np.array([[2, 1],
              [1, 2]])

eigenvalues = np.linalg.eigvals(A)

print("Eigenvalues:", eigenvalues)
```

### **AI Insight: What Does an Eigenvalue Represent During a Matrix Transformation?**

An eigenvalue represents the **scaling factor** applied to its corresponding eigenvector during a matrix transformation.

If the eigenvalue is:

*   Greater than 1, the vector increases in magnitude.
*   Between 0 and 1, the vector decreases in magnitude.
*   Negative, the vector reverses direction.

### **Task 2: Verify the Eigenvectors Using NumPy**

```python
import numpy as np

# Define the given matrix
A = np.array([[2, 1],
              [1, 2]], dtype=float)

# Calculate the eigenvalues and eigenvectors of matrix A
eigenvalues, eigenvectors = np.linalg.eig(A)

# Display the calculated eigenvalues
print("Eigenvalues:")
print(eigenvalues)

# Display the eigenvectors (each column represents one eigenvector)
print("\nEigenvectors (columns):")
print(eigenvectors)

# Verify that A @ v = lambda * v for each eigenvalue and eigenvector
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]       # Get the current eigenvalue
    v = eigenvectors[:, i]     # Get its corresponding eigenvector

    # Display the matrix-vector multiplication result
    print(f"\nEigenvalue: {lam}")
    print("A @ v =", A @ v)

    # Display the eigenvalue multiplied by its eigenvector
    print("lambda * v =", lam * v)

    # Check whether both sides of the eigenvector equation are equal
    assert np.allclose(A @ v, lam * v)

# Print a confirmation message if all checks pass
print("\nAll eigenvectors verified successfully!")
```

# **Mission 5 – Matrix Diagonalization**

### **Task 1: Construct Matrix P**

### **Task 2: Construct Matrix D**

### **Task 3: Verify A = P D P⁻¹ Using NumPy**

مهمة اربعة.png

```python
import numpy as np

# Define the original matrix
A = np.array([[2, 1],
              [1, 2]], dtype=float)

# Construct P using the eigenvectors as columns
P = np.array([[1, 1],
              [1, -1]], dtype=float)

# Construct D using the corresponding eigenvalues
D = np.diag([3, 1])

# Calculate the inverse of P
P_inv = np.linalg.inv(P)

# Reconstruct A using the diagonalization formula
A_reconstructed = P @ D @ P_inv

# Display the matrices and the reconstructed result
print("P =\n", P)
print("D =\n", D)
print("P inverse =\n", P_inv)
print("P @ D @ P inverse =\n", A_reconstructed)

# Verify that the reconstructed matrix equals A
assert np.allclose(A, A_reconstructed)

print("Diagonalization verified successfully!")
```

### **Task 4: Explain Why Diagonal Matrices Are Easier to Work With**

Diagonal matrices are easier to work with because most of their elements are zero. Matrix operations are simpler, and calculating powers of a diagonal matrix only requires raising its diagonal elements to the required power.

### **AI Insight: Why Is Diagonalization Useful in Machine Learning?**

Diagonalization simplifies matrix computations by transforming a matrix into a diagonal form. It helps analyze important directions and scaling factors and is related to PCA, which uses eigenvectors and eigenvalues to identify principal directions in data.
